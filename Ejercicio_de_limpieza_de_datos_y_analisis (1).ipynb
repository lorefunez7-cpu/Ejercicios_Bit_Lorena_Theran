```python
df_limpio =df.copy()
```

```python
# convertir vacios y todo en minuscula
df_limpio["producto"]= df_limpio["producto"].str.lower().str.strip()
df_limpio["pais"]= df_limpio["pais"].str.lower().str.strip()
df_limpio["metodo_pago"]= df_limpio["metodo_pago"].str.lower().str.strip()
```

```python
#estandarizando
df_limpio["pais"]=df_limpio["pais"].replace({
 "perú":"peru",
"Col":"colombia",
"col":"colombia"
})

df_limpio["metodo_pago"]=df_limpio["metodo_pago"].replace({
"TRANSFERENCIA":"transferencia"

})
```

```python
#convertir a numero
df_limpio['cantidad']=pd.to_numeric(df_limpio["cantidad"],errors="coerce")
df_limpio['precio']=pd.to_numeric(df_limpio["precio"],errors="coerce")

#reemplazar celdas vacias con la mediana
df_limpio["precio"]=df_limpio["precio"].fillna(df_limpio["precio"].median())
df_limpio["cantidad"]=df_limpio["cantidad"].fillna(df_limpio["cantidad"].median())
```

```python
#pasar formato de fecha
df_limpio["fecha"]=pd.to_datetime(df_limpio["fecha"],errors="coerce")
df_limpio=df_limpio.dropna(subset=["fecha"])
```

```python
#valores atipicos (outliers)
df_limpio= df_limpio[df_limpio["precio"]<100000]
```

```python
#verificacion final
print("--- Diagnóstico de limpieza completo ---")
print(df_limpio.info())
```

```python
df_limpio.isnull().sum()
```

```python
df_limpio = df_limpio.reset_index(drop=True)
```

**Parte 3 — Análisis con Pandas
Actividades**

• Crear una nueva columna llamada total: df['total'] = df['precio'] * df['cantidad']

Obtener:

• Total vendido

• Promedio de ventas

• Venta máxima

• Venta mínima

**Identificar:**

Los 5 productos con mayor valor vendido
El país con más ventas

```python
df_limpio['total'] = df_limpio['precio'] * df_limpio['cantidad']
print(f"Total vendido: {df_limpio['total'].sum():,.0f}")
promedio = df_limpio['total'].mean()
print(f"Promedio de ventas: {  promedio:.0f}")
maximo = df_limpio['total'].max()
print(f"Venta máxima:{maximo:.0f}")
minima=df_limpio['total'].min()
print(f"Venta mínima:{ minima:.0f}")
```

```python
#agrupar productos
productos_top = df_limpio.groupby('producto')['total'].sum()
productos_top = productos_top.sort_values(ascending=False).head(5)
print("--- TOP 5 PRODUCTOS CON MÁS VENTAS ---")
for producto, valor in productos_top.items():
    print(f"Producto: {producto:<20} Total Vendido: ${valor:,.0f}")
```

```python
#agrupar pais
ventas_por_pais = df_limpio.groupby('pais')['total'].sum()
pais_con_mayor_ventas = ventas_por_pais.idxmax()
monto_maximo = ventas_por_pais.max()
print(f"El país con más ventas es {pais_con_mayor_ventas:} con un monto total de ${monto_maximo:,.0f}")
```

**Parte 4 — Introducción a NumPy
Actividades**

• Convertir columnas numéricas a un array de NumPy: data = df[['precio',
'cantidad']].to_numpy()

• Separar columnas utilizando indexación: precios = data[:, 0]
cantidades = data[:, 1]

• Aplicar vectorización para calcular ventas totales: totales = precios *
cantidades

```python
ventas = df_limpio[['precio', 'cantidad']].to_numpy()
#estraer las columnas
precios = ventas[:, 0]
cantidades = ventas[:, 1]
#multiplicamos
total_ventas = precios * cantidades
print(f"Promedio ventas{ np.mean(total_ventas):.0f}")
print("Ventas mayores a 5000: ", len (total_ventas[total_ventas>5000]))
valor_ventas_mayores = np.sum(total_ventas[total_ventas > 5000])
print(f"Valor total de ventas mayores a 5000: ${valor_ventas_mayores:,.2f}")
```

**Parte 5 — Análisis con NumPy
Utilizando funciones de NumPy, calcular:**

• Suma total de ventas

• Promedio de ventas

• Venta máxima

• Cantidad de ventas superiores a 1000  

```python
print(f"Suma total: ${np.sum(total_ventas):,.0f}")
print(f"Promedio de ventas: ${np.mean(total_ventas):,.0f}")
print(f"Venta maxima: ${np.max(total_ventas):,.0f}")
print(f"Cantidad de ventas superiores a 1000: ${np.sum(total_ventas[total_ventas>1000]):,.0f}")
```
