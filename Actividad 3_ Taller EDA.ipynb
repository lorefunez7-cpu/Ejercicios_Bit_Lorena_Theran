```python
#Filtrado y selección de datos relevantes
#Se realiza una nueva copia  a la base
columnas_clave = ['customerID', 'Contract', 'tenure', 'MonthlyCharges', 'TotalCharges', 'Churn']
df_analisis = df_nueva[columnas_clave].copy()
```

```python
#Creamos un filtro específico para los clientes con contratos de 1 año o más
clientes_contrato_largo = df_analisis[df_analisis['Contract'].isin(['One year', 'Two year'])].reset_index(drop=True)
```

```python
# 1. Transformamos los meses de antigüedad ('tenure') a Años para medir mejor la fidelidad.
df_analisis['Anos_Fidelidad'] = df_analisis['tenure'] / 12

# 2. Creamos una variable para clasificar el nivel de aporte económico del cliente (Segmentación VIP)
# El percentil 75 de TotalCharges ronda los $3,800. Usaremos ese límite para hallar el Top de dinero.
df_analisis['Segmento_Cliente'] = np.where(df_analisis['TotalCharges'] >= 3800, 'VIP (Alto Valor)', 'Estándar')

# 3. Creamos una bandera numérica para medir la rotación de manera matemática fácilmente
df_analisis['Rotado_Numerico'] = np.where(df_analisis['Churn'] == 'Yes', 1, 0)
```

```python
# Asegurar formato numérico en las columnas clave
df_nueva['TotalCharges'] = pd.to_numeric(df_nueva['TotalCharges'], errors='coerce').fillna(0)
df_nueva['MonthlyCharges'] = pd.to_numeric(df_nueva['MonthlyCharges'], errors='coerce').fillna(0)

# =========================================================================
# PASO 1: FILTRADO Y SELECCIÓN DE DATOS RELEVANTES
# =========================================================================
# Seleccionamos las columnas de identificación, contrato, abandono, dinero y los productos principales
columnas_clave = [
    'customerID', 'Contract', 'tenure', 'MonthlyCharges', 'TotalCharges', 'Churn',
    'InternetService', 'OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport'
]
df_analisis = df_nueva[columnas_clave].copy()

# Filtrado A: Clientes fieles de 1 a 3 años (12 a 36 meses de antigüedad)
clientes_1_a_3_anos = df_analisis[(df_analisis['tenure'] >= 12) & (df_analisis['tenure'] <= 36)].reset_index(drop=True)



# PASO 2: TRANSFORMACIÓN DE COLUMNAS / CREACIÓN DE VARIABLES NUEVAS

# Convertimos la antigüedad a años para facilitar la lectura
df_analisis['Anos_Fidelidad'] = df_analisis['tenure'] / 12

# Identificamos de forma numérica la rotación (1 = Se fue, 0 = Se quedó)
df_analisis['Rotado_Numerico'] = np.where(df_analisis['Churn'] == 'yes', 1, 0)

# Alerta de Riesgo (Alta probabilidad de irse):
# El perfil crítico en Telco son clientes "mes a mes", con cargos altos (mayor al promedio de $64) y sin soporte técnico.
perfil_riesgo = (df_analisis['Contract'] == 'month-to-month') & \
                (df_analisis['MonthlyCharges'] > 64.7) & \
                (df_analisis['TechSupport'] == 'no')

df_analisis['Alerta_Abandono'] = np.where(perfil_riesgo, 'Alta Probabilidad', 'Estable')



# PASO 3: AGRUPACIONES, RESÚMENES Y DETALLE (MOSTRANDO TOP 5 EN CADA UNO)


# --- 3(A). Clientes que más rotan y POR QUÉ (Análisis del producto e Internet) ---
# Agrupamos por contrato e internet para ver dónde se concentra la fuga de clientes
analisis_fuga = df_analisis.groupby(['Contract', 'InternetService']).agg(
    Total_Clientes=('customerID', 'count'),
    Clientes_Perdidos=('Rotado_Numerico', 'sum'),
    Tasa_Rotacion_Porc=('Rotado_Numerico', lambda x: x.mean() * 100),
    Cargos_Mensuales_Promedio=('MonthlyCharges', 'mean')
).sort_values(by='Tasa_Rotacion_Porc', ascending=False).reset_index()

# --- 3(B). Clientes que más generan ingresos (Top 5) ---
top_5_ingresos = df_analisis.sort_values(by='TotalCharges', ascending=False).head(5).reset_index(drop=True)

# --- 3(C). Clientes con mayor probabilidad de irse y sus productos (Top 5 en riesgo) ---
# Filtramos los que tienen Alerta Alta, ordenamos por sus cargos mensuales y vemos qué servicios contrataron
top_5_riesgo = df_analisis[df_analisis['Alerta_Abandono'] == 'Alta Probabilidad']\
               .sort_values(by='MonthlyCharges', ascending=False).head(5).reset_index(drop=True)


# =========================================================================
# IMPRESIÓN DEL REPORTE DETALLADO (TOP 5 EN CADA PASO)
# =========================================================================

print("=====================================================================")
print("  PASO 1: CLIENTES FIELES DE 1 A 3 AÑOS (MUESTRA DE LOS PRIMEROS 5)")
print("=====================================================================")
print(f"Total de clientes en este rango de antigüedad: {len(clientes_1_a_3_anos)}\n")
print(clientes_1_a_3_anos[['customerID', 'Contract', 'tenure', 'MonthlyCharges', 'TotalCharges']].head(5).to_string(index=False))


print("\n" + "="*75)
print("  PASO 3(A): ¿QUIÉNES ROTAN MÁS Y POR QUÉ? (ANÁLISIS DE CONTRATO Y PRODUCTO)")
print("="*75)
print("Nota: Aquí verás las combinaciones de producto que causan que el cliente NO sea fiel:")
print(analisis_fuga.head(5).to_string(index=False))


print("\n" + "="*75)
print("  PASO 3(B): TOP 5 CLIENTES QUE MÁS GENERAN INGRESOS")
print("="*75)
top_5_ingresos_visual = top_5_ingresos[['customerID', 'Contract', 'tenure', 'MonthlyCharges', 'TotalCharges']].copy()
top_5_ingresos_visual['TotalCharges'] = top_5_ingresos_visual['TotalCharges'].map(lambda x: f"${x:,.2f}")
print(top_5_ingresos_visual.to_string(index=False))


print("\n" + "="*75)
print("  PASO 3(C): TOP 5 CLIENTES CON ALTA PROBABILIDAD DE IRSE Y SUS PRODUCTOS")
print("="*75)
columnas_riesgo_productos = ['customerID', 'Contract', 'MonthlyCharges', 'InternetService', 'TechSupport', 'OnlineSecurity']
print(top_5_riesgo[columnas_riesgo_productos].to_string(index=False))
```

```python
# 1. Asegurar formato numérico en las columnas de dinero
df_nueva['TotalCharges'] = pd.to_numeric(df_nueva['TotalCharges'], errors='coerce').fillna(0)
df_nueva['MonthlyCharges'] = pd.to_numeric(df_nueva['MonthlyCharges'], errors='coerce').fillna(0)

# =========================================================================
# PASO 1: FILTRADO Y SELECCIÓN RELEVANTE (Los datos que nos importan)
# =========================================================================
columnas_clave = [
    'customerID', 'Contract', 'tenure', 'MonthlyCharges', 'TotalCharges', 'Churn',
    'InternetService', 'TechSupport', 'Partner', 'Dependents'
]
df_limpio = df_nueva[columnas_clave].copy()

# Filtrado solicitado: Clientes estables de 1 a 3 años (12 a 36 meses)
clientes_1_3_anos = df_limpio[(df_limpio['tenure'] >= 12) & (df_limpio['tenure'] <= 36)]


# =========================================================================
# PASO 2: TRANSFORMACIÓN (Traducción a conceptos humanos)
# =========================================================================
# Convertir meses a años
df_limpio['Anos_Fidelidad'] = df_limpio['tenure'] / 12

# Identificar clientes VIP (Los que han pagado más de $3,800 en total)
df_limpio['Es_VIP'] = np.where(df_limpio['TotalCharges'] >= 3800, 'VIP', 'Estándar')

# Activar alerta roja para clientes con alta probabilidad de irse
# Perfil: Contrato mes a mes, internet de fibra óptica y sin soporte técnico
clientes_en_peligro = df_limpio[
    (df_limpio['Contract'] == 'month-to-month') &
    (df_limpio['InternetService'] == 'fiber optic') &
    (df_limpio['TechSupport'] == 'no')
]


# =========================================================================
# PASO 3: AGRUPACIONES Y REPORTES EN ESPAÑOL DIRECTO
# =========================================================================

# Cálculos rápidos para el reporte
total_clientes = len(df_limpio)
clientes_mes_a_mes = len(df_limpio[df_limpio['Contract'] == 'month-to-month'])
fuga_mes_a_mes = (df_limpio[df_limpio['Contract'] == 'month-to-month']['Churn'] == 'yes').mean() * 100

clientes_vip = df_limpio[df_limpio['Es_VIP'] == 'VIP']
dinero_vip = clientes_vip['TotalCharges'].sum()
dinero_total = df_limpio['TotalCharges'].sum()

promedio_solteros = df_limpio[(df_limpio['Partner'] == 'no') & (df_limpio['Dependents'] == 'no')]['tenure'].mean()
promedio_familias = df_limpio[(df_limpio['Partner'] == 'yes') & (df_limpio['Dependents'] == 'yes')]['tenure'].mean()

# =========================================================================
# IMPRESIÓN DEL DIAGNÓSTICO EN LENGUAJE NATURAL
# =========================================================================
print("=====================================================================")
print("     ¿QUÉ LE PASA A NUESTRO NEGOCIO?         ")
print("=====================================================================\n")

print(f"1. EL PELIGRO DEL 'MES A MES':")
print(f"   * Tienes {clientes_mes_a_mes} clientes sin contrato a largo plazo.")
print(f"   * ¡El {fuga_mes_a_mes:.1f}% de ellos se va de la empresa! Son tu mayor fuga de dinero.\n")

print(f"2. ¿DÓNDE ESTÁ LA MINA DE ORO? (INGRESOS):")
print(f"   * Tienes {len(clientes_vip)} clientes catalogados como VIP.")
print(f"   * Aunque son poquitos, aportan ${dinero_vip:,.2f} dólares.")
print(f"   * ¡Eso es el {(dinero_vip/dinero_total)*100:.1f}% de TODO el dinero de la compañía!\n")

print(f"3. LAS FAMILIAS SÍ SON FIELES:")
print(f"   * Un cliente soltero y sin hijos dura en promedio apenas {promedio_solteros:.1f} meses.")
print(f"   * Un cliente con pareja e hijos dura en promedio {promedio_familias:.1f} meses (¡Casi el doble!).\n")

print(f"4. ALERTA ROJA (Clientes con alta probabilidad de irse hoy mismo):")
print(f"   * Detectamos {len(clientes_en_peligro)} clientes críticos en la base de datos.")
print(f"   * ¿Su perfil? Tienen internet caro (Fibra Óptica), pagan mes a mes y NO tienen Soporte Técnico.")
print(f"   * Aquí tienes el ID de los primeros 5 clientes que deberías llamar para salvarlos:")
print(clientes_en_peligro[['customerID', 'MonthlyCharges']].head(5).to_string(index=False))

print("\n=====================================================================")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configurar el estilo visual de las gráficas
sns.set_theme(style="whitegrid")
plt.rcParams['font.family'] = 'sans-serif'

# Crear el lienzo con 4 subgráficas (2 filas y 2 columnas)
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('DIAGNÓSTICO VISUAL DEL NEGOCIO (MÉTRICAS CLAVE)', fontsize=20, fontweight='bold', color='#112233')

# -------------------------------------------------------------------------
# GRÁFICA 1: El peligro del contrato "Mes a Mes" (Gráfico de Pastel)
# -------------------------------------------------------------------------
porcentajes_m2m = [fuga_mes_a_mes, 100 - fuga_mes_a_mes]
etiquetas_m2m = ['Se Van (Fuga)', 'Se Quedan']

axes[0, 0].pie(porcentajes_m2m, labels=etiquetas_m2m, autopct='%1.1f%%',
               startangle=90, colors=['#e74c3c', '#95a5a6'],
               textprops={'fontweight': 'bold', 'fontsize': 11}, explode=(0.05, 0))
axes[0, 0].set_title('1. Comportamiento Clientes "Mes a Mes"', fontsize=14, fontweight='bold', pad=15)


# -------------------------------------------------------------------------
# GRÁFICA 2: Participación en los Ingresos de Clientes VIP (Gráfico de Barras)
# -------------------------------------------------------------------------
dinero_estandar = dinero_total - dinero_vip
ingresos_valores = [dinero_vip, dinero_estandar]
ingresos_etiquetas = ['Segmento VIP', 'Segmento Estándar']

bars = axes[0, 1].bar(ingresos_etiquetas, ingresos_valores, color=['#f1c40f', '#3498db'], width=0.5)
axes[0, 1].set_title('2. Distribución de Ingresos Totales ($)', fontsize=14, fontweight='bold', pad=15)
axes[0, 1].set_ylabel('Total Acumulado ($)', fontsize=12)

# Añadir etiquetas de dinero sobre las barras para leerlo fácil
for bar in bars:
    yval = bar.get_height()
    axes[0, 1].text(bar.get_x() + bar.get_width()/2, yval + (dinero_total*0.01),
                    f"${yval:,.2f}", ha='center', va='bottom', fontweight='bold')


# -------------------------------------------------------------------------
# GRÁFICA 3: Fidelidad de Familias vs Solteros (Gráfico de Barras Horizontal)
# -------------------------------------------------------------------------
categorias_familia = ['Familias\n(Con Pareja e Hijos)', 'Solteros\n(Sin Familia)']
meses_promedio = [promedio_familias, promedio_solteros]

axes[1, 0].barh(categorias_familia, meses_promedio, color=['#2ecc71', '#e67e22'], height=0.5)
axes[1, 0].set_title('3. Antigüedad Promedio por Núcleo Familiar', fontsize=14, fontweight='bold', pad=15)
axes[1, 0].set_xlabel('Meses de Permanencia en la Empresa', fontsize=12)
axes[1, 0].set_xlim(0, max(meses_promedio) + 10)

# Añadir el número exacto al final de cada barra horizontal
for index, value in enumerate(meses_promedio):
    axes[1, 0].text(value + 1, index, f"{value:.1f} meses", va='center', fontweight='bold')


# -------------------------------------------------------------------------
# CUADRO 4: Alerta Roja (Resumen Informativo)
# -------------------------------------------------------------------------
axes[1, 1].axis('off') # Ocultamos los ejes para usarlo como panel de texto
texto_alerta = (
    f"4. ALERTA ROJA: RIESGO DE ABANDONO\n\n"
    f"Se detectaron {len(clientes_en_peligro)} clientes en riesgo crítico.\n"
    f"Perfil común: Pago mes a mes, Fibra Óptica y sin soporte técnico.\n\n"
    f"Top 5 IDs de clientes a contactar de inmediato:\n\n"
    f"{clientes_en_peligro[['customerID', 'MonthlyCharges']].head(5).to_string(index=False)}"
)
axes[1, 1].text(0.05, 0.5, texto_alerta, fontsize=12, color='#c0392b',
                fontweight='bold', family='monospace', va='center',
                bbox=dict(boxstyle="round,pad=1.5", facecolor='#fdf2f2', edgecolor='#f5c6cb'))

# Ajustar los espacios entre las gráficas para que no se encimen
plt.tight_layout()
plt.show()
```
